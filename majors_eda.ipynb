Analiza wartości institution_selectivity_pctile według poziomu instytucji wykazała wyraźnie rozdzielone zakresy dla poszczególnych kategorii. Zmienna prawdopodobnie pełni funkcję syntetycznego wskaźnika selektywności, mimo nazwy sugerującej klasyczny percentyl.

```python
roi_by_category = (
    df.groupby("major_category")
      .agg(
          count=("net_roi_usd", "size"),
          median_roi=("net_roi_usd", "median"),
          mean_roi=("net_roi_usd", "mean"),
          negative_roi_share=("net_roi_usd", lambda x: (x < 0).mean())
      )
      .sort_values("negative_roi_share", ascending=False)
)

roi_by_category
```

Analiza według kategorii kierunków pokazuje wyraźne różnice w wartości net_roi_usd. Kategorie arts i humanities wyróżniają się ujemną zarówno średnią, jak i medianą ROI, a większość obserwacji w tych grupach osiąga wynik poniżej zera. W przypadku social_science ujemne ROI występuje relatywnie często, jednak mediana i średnia pozostają dodatnie, co sugeruje, że wcześniejszy zaobserwowany przypadek nie jest reprezentatywny dla całej kategorii.

```python
roi_by_major = (
    df.groupby("major")
      .agg(
          count=("net_roi_usd", "size"),
          median_roi=("net_roi_usd", "median"),
          negative_roi_share=("net_roi_usd", lambda x: (x < 0).mean())
      )
      .sort_values("negative_roi_share", ascending=False)
)

roi_by_major.head(15)
```

```python
df.groupby("major_category")["major"].unique()
```

Sprawdzenie struktury kategorii wykazało, że część z nich obejmuje tylko jeden kierunek, m.in. arts, humanities, education i health. W ich przypadku analiza na poziomie kierunku nie stanowi dodatkowego potwierdzenia wyników kategorii. Większą wartość analityczną ma porównanie kierunków wewnątrz bardziej zróżnicowanych kategorii, takich jak social_science, business i STEM.

```python
negative_roi_by_major = (
    df.groupby('major')['net_roi_usd']
    .apply(lambda x: (x < 0).mean() * 100)
    .sort_values(ascending=True)
    .reset_index(name='negative_roi_pct')
)
```

```python
plt.figure(figsize=(10, 8))
bars = plt.barh(
    negative_roi_by_major['major'],
    negative_roi_by_major['negative_roi_pct'],
    color='#d95f02'
)

# Formatowanie osi i tytułów
plt.title('Udział ujemnego ROI dla wszystkich kierunków studiów (%)', fontsize=13, fontweight='bold', pad=15)
plt.xlabel('Odsetek absolwentów z ujemnym ROI (%)', fontsize=11)
plt.ylabel('Kierunek studiów', fontsize=11)
plt.xlim(0, 100)
plt.grid(axis='x', linestyle='--', alpha=0.5)

# Dodanie etykiet z dokładnymi wartościami przy słupkach
for bar in bars:
    width = bar.get_width()
    plt.text(
        width + 1,
        bar.get_y() + bar.get_height() / 2,
        f'{width:.1f}%',
        va='center',
        ha='left',
        fontsize=9
    )

plt.tight_layout()
plt.show()
```

Występowanie ujemnego ROI jest silnie zróżnicowane pomiędzy kierunkami. Najwyższy odsetek obserwacji z ujemnym ROI występuje dla fine_arts (88,8%) oraz humanities (62,3%). Relatywnie wysoki udział widoczny jest również dla psychology, social_work i education, gdzie ujemny ROI dotyczy około 42–48% obserwacji. W przypadku kierunków biznesowych, technicznych i medycznych odsetek ten jest znacznie niższy, a dla części kierunków nie przekracza 1%. Wyniki wskazują więc, że ujemny ROI nie jest równomiernie rozłożony w zbiorze, lecz wyraźnie koncentruje się w określonych kierunkach. Jest to jednak zależność obserwowana w analizowanym zbiorze danych i sama w sobie nie pozwala stwierdzić, że wybór danego kierunku jest przyczyną niższego ROI.
