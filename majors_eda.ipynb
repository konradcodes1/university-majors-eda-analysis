Wartości powyżej 100 występują wyłącznie dla instytucji oznaczonych jako ivy_elite. Obejmują 288 obserwacji, czyli 0,96% zbioru, i przyjmują wyłącznie wartości 101 lub 102. Sugeruje to, że przekroczenie standardowego zakresu percentyla może wynikać ze sposobu konstrukcji zmiennej, a nie z przypadkowych błędów danych.

```python
df.groupby("institution_tier")["institution_selectivity_pctile"].agg(
    ["min", "max", "median"]
)
```

```python
df[df["institution_tier"] == "ivy_elite"][
    "institution_selectivity_pctile"
].value_counts().sort_index()
```

Analiza wartości institution_selectivity_pctile według poziomu instytucji wykazała wyraźnie rozdzielone zakresy dla poszczególnych kategorii. Wartości 101–102 występują w ramach spójnego zakresu 90–102 dla ivy_elite, dlatego nie zostały uznane za błędne obserwacje. Zmienna prawdopodobnie pełni funkcję syntetycznego wskaźnika selektywności, mimo nazwy sugerującej klasyczny percentyl.

```python
roi_by_category = (
    df.groupby("major_category")
      .agg(
          count=("net_roi_usd", "size"),
          median_roi=("net_roi_usd", "median"),
          mean_roi=("net_roi_usd", "mean"),
          negative_roi_share=("net_roi_usd", lambda x: (x < 0).mean())
      )
      .sort_values("negative_roi_share", ascending=False)
)

roi_by_category
```

Analiza według kategorii kierunków pokazuje wyraźne różnice w wartości net_roi_usd. Kategorie arts i humanities wyróżniają się ujemną zarówno średnią, jak i medianą ROI, a większość obserwacji w tych grupach osiąga wynik poniżej zera. W przypadku social_science ujemne ROI występuje relatywnie często, jednak mediana i średnia pozostają dodatnie, co sugeruje, że wcześniejszy zaobserwowany przypadek nie jest reprezentatywny dla całej kategorii.

```python
roi_by_major = (
    df.groupby("major")
      .agg(
          count=("net_roi_usd", "size"),
          median_roi=("net_roi_usd", "median"),
          negative_roi_share=("net_roi_usd", lambda x: (x < 0).mean())
      )
      .sort_values("negative_roi_share", ascending=False)
)

roi_by_major.head(15)
```
